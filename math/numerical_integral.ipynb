# Second integral

```python
f2_star = (4/theta^2) * exp( (-2/theta)*( tau2-tau1 + 3*(t1-tau2) + t2-t1   ) )

pretty_print( diff(f2_star, t1)/f2_star )
pretty_print( diff(f2_star, t2)/f2_star )
pretty_print( diff(f2_star, tau1)/f2_star )
pretty_print( diff(f2_star, tau2)/f2_star )
```

```python
def f2_deriv(self, *args, **kwds):
    param_idx = kwds['diff_param']
    if param_idx == 0:
        return -4 / theta * self(*args)
    elif param_idx == 1:
        return -2 / theta * self(*args)
    elif param_idx == 2:
        return 2 / theta * self(*args)
    else:
        return 4 / theta * self(*args)
        
f2 = function("f_2", nargs=4, derivative_func=f2_deriv)
```

```python
I2 = integral(
    integral(
        P2(t1,t2)*f2(t1,t2,tau1,tau2), 
        (t1,tau2,t2)
    ),
    (t2, tau2, oo),
)

pretty_print(I2)
```

## $\frac{d I_2}{d \tau_1}$

```python
pretty_print(
    diff(I2, tau1)
)
```

```python
pretty_print(
    diff(I2, tau1) - (2/theta)*I2
)
```
